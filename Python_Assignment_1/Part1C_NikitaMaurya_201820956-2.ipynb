```python
a = input("Enter side 1: ")
b = input("Enter side 2: ")
c = input("Enter side 3: ")
l = []
if a.isdigit() and b.isdigit() and c.isdigit(): #check if all the entries are positive integers or not 
    a = int(a) #if yes then convert them to integers
    b = int(b)
    c = int(c)
    if a >= 1 and a <= 1000 and b >= 1 and b <= 1000 and c >= 1 and c <= 1000: #to check if they are between 1 to 1000
        l.append(a) #adding to the list to sort the side which simply the further analysis  
        l.append(b)
        l.append(c)
        print("Entered values:",l)
        l.sort() #to sort the sides
        print("Sorted values: ",l)
        A = l[0] 
        B = l[1]
        C = l[2]
        if A + B <= C: #first to check it is even a triangle or not, or why to do so much mehnat (mehnat is effort in hindi language)
            print("Triangle is impossible")
        elif A == B == C: #then simply check is all sides are equal or not
            print("Triangle is equilateral")
        elif C**2 == (A**2 + B**2): #check if it a right angled or not 
            print("Triangle is right angled")
        elif A == B or B == C or A == C: #if it is not right angled then check if it isoceles because all right angled are either isoceles or scalene
            print("Triangle is isoceles")
        elif A != B and B != C and A != C: #then check for scalene
            print("Triangle is scalene")
    else:
        print("Invalid: All of the sides need to be within the range 1-1000") #to handle error out of range
else:
    print("Invalid: All of the sides need to be within the range 1-1000") #to handle error if entries are not numbers
```
