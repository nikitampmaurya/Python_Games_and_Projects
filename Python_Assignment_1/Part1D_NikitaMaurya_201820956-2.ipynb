```python
l = [10, 78, 65, 96, 76, -27, 100, 191, 56, 79, 60, 33, 
     56, 50, 75, 45, 99, 60, 67, 68, 31, 99, 21, 0, 55, 
     76, 45, 25, 78, 78, 80, 59, 0, 24, 66, 24, 103, 12] #given list

D = []
M = []
P = []
F = []
E = []

for i in range(len(l)):
    if l[i] >= 70 and l[i] <= 100:    #checking and adding score to distinction list 
        D.append(l[i])
    elif l[i] >= 60 and l[i] <= 69:   #checking and adding score to merit list
        M.append(l[i])
    elif l[i] >= 40 and l[i] <= 59:   #checking and adding score to pass list
        P.append(l[i])
    elif l[i] >= 0 and l[i] <= 39:    #checking and adding score to fail list
        F.append(l[i]) 
    else:
        E.append(i) #adding an index of the errors

D.sort(reverse=True) #largest to smallest
M.sort(reverse=True) #largest to smallest
P.sort(reverse=True) #largest to smallest
F.sort() #smallest to largest


Dm = round(sum(D)/len(D),2) #rounding up to two deciaml points 
Mm = round(sum(M)/len(M),2)
Pm = round(sum(P)/len(P),2)
Fm = round(sum(F)/len(F),2)

print(f"Distinction (mean={Dm}): ", end="") #to continue the next line on the same line
for i in D:
    print(i, end =" ") #to write all numbers in one line with space
print()

print(f"Merit (mean={Mm}): ", end="")
for i in M:
    print(i, end=" ")
    
print(f"\nPass (mean={Pm}): ", end="")
for i in P:
    print(i, end=" ")

print(f"\nFail (mean={Fm}): ", end="")
for i in F:
    print(i, end= " ")

print("\nErrors in position: ", end="")
for i in E:
    print(i, end = " ")
```
