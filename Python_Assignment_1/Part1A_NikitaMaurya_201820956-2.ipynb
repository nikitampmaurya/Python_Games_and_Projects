```python
while True: #to keep asking for dna sequence until the given conditions are satisfied 
    
    dna1 = input("Hi there, please enter your first dna sequence: ") # Step 1: Asked for first DNA sequence
    
    if dna1.isalpha(): #check if the given input has only alphabets 
        
        print("Thank you entering your first dna sequence!")
        
        dna2 = input("Now, please enter your second dna sequence: ") # Step 2: Asked for second DNA sequence
        
        if dna2.isalpha(): #check again if the second input has only alphabets 
            
            print("Thank you entering your second dna sequence")
            
            dna = dna1+dna2  # Step 3: Joint both the DNA sequences  
            
            DNA = dna.upper() # Step 4: Converted them to uppercase
            
            print("This is your new DNA sequence: ",DNA) 
            
            print("Length of your new DNA: ",len(DNA)) # Step 5: calculated the length of joint DNA

            # Step 6: for reverse complement -  

            # first I reversed the DNA sequence from left to right --> right to left

            # then I created the complement for the reversed dna 
            
            list1 = list(DNA) # split nucleotides into a list 
            
            rev_list1 = list1[::-1] # reverse dna  
            
            rev_dna = "" #create an empty sequence
            
            for i in rev_list1: # add nucleotides to the empty sequence after reversing it
                rev_dna += i
            
            #print(rev_dna) 
            
            rev_comp = "" #create another empty sequence for creating a complement of the reversed dna
            
            for i in rev_dna:
                if i == "A":
                    rev_comp += "T"
                elif i == "T":
                    rev_comp += "A"
                elif i == "G":
                    rev_comp += "C"
                elif i == "C":
                    rev_comp += "G"
                else:
                    rev_comp += "i"
                    
            print("Reverse Complement of your DNA: ",rev_comp)

            #Step 7: Convert the joint DNA into RNA

            RNA = "" #creating empty rna sequence

            for i in DNA:
                if i == "T":
                    RNA += "U" #replacing T with U
                else:
                    RNA += i

            polytailA = "AAAAAAA"
            
            new_rna = RNA + polytailA
            
            print("Length of new RNA: ",len(new_rna)) #Step 8: calculated the length of RNA
            
            print("RNA: ", new_rna)
            
            break
            
        else:
            print("oops..! try again")
    else:
        print("oops..! try again")
```
