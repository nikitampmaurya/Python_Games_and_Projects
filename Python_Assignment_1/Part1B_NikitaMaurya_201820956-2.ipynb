```python
import math #for taking squareroot of variance 
while True: #keeps asking for valid input until the conditions are satisfied.
    a = input("Please enter value of a: ")
    if a.isdigit(): #to check if the given in input is a positive integer
        a = int(a) #if the integer is positive then convert it to an integer 
        b = input("Please enter value of b: ")
        if b.isdigit():
            b = int(b)
            c = input("Please enter value of c: ")
            if c.isdigit():
                c = int(c)
                d = input("Please enter value of d: ")
                if d.isdigit():
                    d = int(d)
                    e = input("please enter value of e: ")
                    if e.isdigit():
                        e = int(e)
                        break #once all conditions are satisfied
                    else:
                        print("Oops Invalid! Please enter only positive integers for e.")
                else:
                    print("Oops Invalid! Please enter only positive integers for d.")
            else:
                print("Oops Invalid! Please enter only positive integers for c.")
        else:
            print("Oops Invalid! Please enter only positive integers for b.")
    else:
        print("Oops Invalid! Please enter only positive integers for a.")

#Step 2: Calculate total and mean
L = [a,b,c,d,e]
print("You have entered: ",L)
s = a + b + c + d + e 
mean = s/len(L)
print("Mean: ",mean, "(total = ", s, ", number of variables = 5)", )

#Step 3: Calculate Variance 

D1 = []
sum_of_difference = 0
for i in L:
    v = (i - mean)**2 #calculate the difference between each variable and mean
    D1.append(v)
    sum_of_difference += v #then sum the difference
    
V = sum_of_difference / len(D1) 
print("Variance: ",V)

#Step 4: Calculate Standard Deviation
S = math.sqrt(V)
print("Standard deviation: ",S)
```
